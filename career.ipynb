```python
df['experience_years']=df['experience_years'].str.replace("years","").astype('Int64')
df['experience_years']
```

```python
l = ['age','education_level','field_of_study', 'experience_years', 'openness', 'conscientiousness', 'extraversion', 'agreeableness', 'neuroticism', 
     'technical_skill', 'data_reasoning_skill', 'communication_skill', 'problem_solving_skill', 'leadership_skill', 'creative_thinking_skill',
     'interest_technology', 'interest_business', 'interest_creative', 'interest_health_social', 'interest_research_academic', 'preferred_work_style', 'preferred_environment', 'risk_tolerance']
for i in l:
    print(i, ":", df[i].unique(), "\n")
```

```python
median_cols = ['age', 'experience_years', 'openness', 'conscientiousness','extraversion', 'agreeableness', 'neuroticism',
    'technical_skill', 'data_reasoning_skill', 'communication_skill','problem_solving_skill', 'leadership_skill', 'creative_thinking_skill',
    'interest_technology', 'interest_business', 'interest_creative','interest_health_social', 'interest_research_academic']
for col in median_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
mode_cols = ['education_level', 'field_of_study','preferred_work_style', 'preferred_environment','risk_tolerance']
for col in mode_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns
outlier_len = {}
for i in num_cols:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3 - Q1 
    outliers = df[(df[i] < Q1 - 1.5*IQR) | (df[i] > Q3 + 1.5*IQR)]
    outlier_len[i] = len(outliers)
outlier_len
```

```python
from scipy.stats import chi2_contingency
target = "career_cluster"
categorical_col=df.select_dtypes(include=['object']).columns
results = []
for col in categorical_col:
    contingency_table = pd.crosstab(df[col], df[target])
    chi2, p, dof, expected = chi2_contingency(contingency_table)
    results.append({ "feature": col,"chi_square": round(chi2, 3), "p_value": round(p, 5), "significant_at_0.05": p < 0.05})
cs_df = pd.DataFrame(results).sort_values("p_value")
cs_df
```

```python
df["work_life_alignment"] = (
        df["preferred_environment"].eq("flexible").astype(int)
        -df["preferred_environment"].eq("structured").astype(int) +
        df["risk_tolerance"].eq(0).astype(int) -
        df["risk_tolerance"].eq(2).astype(int))
df.drop(columns=['preferred_environment','risk_tolerance'],inplace=True)
```

```python
categorical_col=df.select_dtypes(include=['object']).columns
encode_col=categorical_col.drop('career_cluster')
df=pd.get_dummies(df,columns=encode_col,drop_first=True)
```

```python
counts = df['career_cluster'].value_counts()
percent = df['career_cluster'].value_counts(normalize=True) * 100
imbalance_df = pd.DataFrame({'count': counts,'percentage': percent.round(2)})
imbalance_df
```

```python
imbalance_ratio = counts.max() / counts.min()
imbalance_ratio
```

```python
x=df.drop(columns='career_cluster')
y=df['career_cluster']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,stratify=y,random_state=42)
x_train
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(
    n_estimators=600,max_depth=9,min_samples_leaf=6,max_features=0.3,          
    ccp_alpha=0.0005,random_state=42,n_jobs=-1)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score
score=accuracy_score(y_test,y_pred)
print(score)
```

```python
train_preds = model.predict(x_train)
test_preds = model.predict(x_test)
train_acc = accuracy_score(y_train, train_preds)
test_acc = accuracy_score(y_test, test_preds)
print(f"Training Accuracy: {train_acc * 100:.2f}%")
print(f"Testing Accuracy: {test_acc * 100:.2f}%")
```

```python
report=classification_report(y_test,y_pred)
print(report)
```

```python
from sklearn.metrics import confusion_matrix
class_names = model.classes_
cm = confusion_matrix(y_test,y_pred)
plt.figure(figsize=(10,8))
sns.heatmap(cm,annot=True,cmap="Blues",fmt="d",xticklabels=class_names,yticklabels=class_names)
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
importance = pd.Series(model.feature_importances_,index=x_train.columns)
top_features = importance.nlargest(15).sort_values()
plt.figure(figsize=(10, 6))
plt.barh(top_features.index,top_features.values)
plt.title("Top 15 Features")
plt.xlabel("Score")
plt.show()
```
